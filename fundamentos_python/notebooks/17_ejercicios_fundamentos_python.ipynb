### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí
inventario_total = {}

inventarios = [inventario_norte, inventario_sur, inventario_centro]

articulos = set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys(),
])

print(articulos)
print("="*50)

for articulo in articulos:
    inventario_total[articulo] = {
        "cantidad_total": sum([inventario.get(articulo, 0) for inventario in inventarios]),
        "sucursales": sum([1 for inventario in inventarios if articulo in inventario])
    }

for articulo in inventario_total:
    print(f"articulo: {articulo} | {inventario_total[articulo]}")
    print('='*50)
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
# Escribe tu solución aquí
agrupados = {}

for empleado in empleados:
    departamento = empleado["departamento"]
    nivel = empleado["nivel"]

    if departamento not in agrupados:
        agrupados[departamento] = {}

    if nivel not in agrupados[departamento]:
        agrupados[departamento][nivel] = {
            "nombres": [],
            "cantidad": 0,
            "salarios": []
        }

    grupo = agrupados[departamento][nivel]
    grupo["nombres"].append(empleado["nombre"])
    grupo["cantidad"] += 1
    grupo["salarios"].append(empleado["salario"])

for departamento in agrupados:
    for nivel in agrupados[departamento]:
        grupo = agrupados[departamento][nivel]
        grupo["salario_promedio"] = sum(grupo["salarios"]) / grupo["cantidad"]
        del grupo["salarios"]

print(agrupados)
```

### Pregunta 3.4: Sistema de votación con validación

Procesa una lista de votos y separa los votos **válidos** de los **inválidos**.

**Requisitos:**
- Los votos deben ser insensibles a mayúsculas (`"ANA"` cuenta para `"Ana"`).
- Registra el conteo por candidato, los votos inválidos y el porcentaje obtenido (sobre los votos válidos).
- Determina si hay un ganador o un empate.

```python
candidatos = {"Ana", "Luis", "Carla"}
votos = ["Ana", "Luis", "ana", "Carla", "Pedro", "Luis", "ANA", "", "Carla"]
```

```python
# Escribe tu solución aquí
conteo = {candidato: 0 for candidato in candidatos}
invalidos = 0

for voto in votos:
    voto_normalizado = voto.strip().lower()

    candidatos_normalizados = {
        candidato.lower(): candidato for candidato in candidatos
    }

    if voto_normalizado in candidatos_normalizados:
        candidato = candidatos_normalizados[voto_normalizado]
        conteo[candidato] += 1
    else:
        invalidos += 1

total_validos = sum(conteo.values())

porcentajes = {}

for candidato, cantidad in conteo.items():
    porcentajes[candidato] = cantidad / total_validos * 100 if total_validos else 0

max_votos = max(conteo.values())

ganadores = [
    candidato for candidato, cantidad in conteo.items()
    if cantidad == max_votos
]

if len(ganadores) == 1:
    resultado = f"Ganador: {ganadores[0]}"
else:
    resultado = f"Empate: {', '.join(ganadores)}"

print("Conteo:", conteo)
print("Votos inválidos:", invalidos)
print("Porcentajes:", porcentajes)
print(resultado)
```

### Pregunta 3.5: Índice invertido de documentos

Crea un **índice invertido** que permita saber en qué documentos aparece cada palabra.

**Requisitos:**
- Cada palabra debe apuntar a un **conjunto** (`set`) de identificadores de documentos.
- Ignora las palabras de una sola letra.
- Usa el índice para responder: ¿en qué documentos aparecen `"python"` y `"datos"`?

```python
documentos = {
    "doc1": "python y ciencia de datos",
    "doc2": "python para desarrollo web",
    "doc3": "ciencia de datos con inteligencia artificial",
}
```

```python
# Escribe tu solución aquí
indice = {}

for documento, texto in documentos.items():
    palabras = texto.lower().split()

    for palabra in palabras:
        if len(palabra) > 1:
            if palabra not in indice:
                indice[palabra] = set()

            indice[palabra].add(documento)

print("Índice invertido:")
print(indice)

print("Documentos con 'python':", indice.get("python", set()))
print("Documentos con 'datos':", indice.get("datos", set()))
```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(ventas: list[float]) -> dict[str, float] | str:
    if not ventas:
        return "ERROR: No existen ventas"

    for venta in ventas:
        if venta < 0:
            return f"ERROR: Valor de venta negativo: {venta}"

    total_vendido = sum(ventas)
    promedio_vendido = total_vendido / len(ventas)
    venta_max = max(ventas)
    venta_min = min(ventas)

    ventas_superiores_promedio = sum([1 for venta in ventas if venta >= promedio_vendido])
    ventas_por_debajo_100 = sum([1 for venta in ventas if venta < 100])

    estadisticos = {
        "total_vendido": total_vendido,
        "promedio_vendido": promedio_vendido,
        "venta_max": venta_max,
        "venta_min": venta_min,
        "ventas_superiores": ventas_superiores_promedio,
        "ventas_por_debajo_100": ventas_por_debajo_100
    }

    return estadisticos


stats = analizar_ventas(ventas)
print(stats)
```

### Pregunta 5.2: Validar una contraseña y explicar los errores

Crea una función `validar_password(password)` que determine si una contraseña cumple todas las reglas de seguridad.

**Requisitos:**
- Debe tener al menos 8 caracteres.
- Debe contener una letra mayúscula, una minúscula y un número.
- Debe incluir al menos uno de estos símbolos: `@`, `#`, `$` o `%`.
- No debe contener espacios.
- Debe retornar una tupla con el formato `(es_valida, errores)`, donde `errores` sea una lista.

```python
password_prueba = 'Python123'
```

```python
# Escribe tu solución aquí
def validar_password(password):
    errores = []

    if len(password) < 8:
        errores.append("Debe tener al menos 8 caracteres")

    if not any(caracter.isupper() for caracter in password):
        errores.append("Debe contener una letra mayúscula")

    if not any(caracter.islower() for caracter in password):
        errores.append("Debe contener una letra minúscula")

    if not any(caracter.isdigit() for caracter in password):
        errores.append("Debe contener un número")

    if not any(caracter in "@#$%" for caracter in password):
        errores.append("Debe contener al menos uno de estos símbolos: @, #, $, %")

    if " " in password:
        errores.append("No debe contener espacios")

    es_valida = len(errores) == 0

    return es_valida, errores

resultado = validar_password(password_prueba)
print(resultado)
```

```python
# Escribe tu solución aquí
def evaluar_empleados(empleados, minimo=70):
    resultados = []

    for nombre, puntajes in empleados.items():
        promedio = sum(puntajes) / len(puntajes)

        if promedio >= 90:
            clasificacion = "Excelente"
        elif promedio >= 80:
            clasificacion = "Bueno"
        elif promedio >= 70:
            clasificacion = "Regular"
        else:
            clasificacion = "Deficiente"

        if promedio >= minimo:
            resultados.append((nombre, promedio, clasificacion))

    def obtener_nombre(resultado):
        return resultado[0]

    def obtener_promedio(resultado):
        return resultado[1]

    resultados.sort(key=obtener_nombre)
    resultados.sort(key=obtener_promedio, reverse=True)

    return resultados

resultado = evaluar_empleados(empleados)
print(resultado)
```

### Pregunta 5.4: Comprimir y descomprimir texto

Implementa dos funciones: `comprimir_texto(texto)` y `descomprimir_texto(texto)`.

**Requisitos:**
- `comprimir_texto` debe reemplazar caracteres consecutivos repetidos por el carácter y su cantidad.
  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaabbccccdaa'
```

```python
# Escribe tu solución aquí
def comprimir_texto(texto):
    if not texto:
        return ""

    resultado = []
    contador = 1

    for i in range(1, len(texto)):
        if texto[i] == texto[i - 1]:
            contador += 1
        else:
            resultado.append(texto[i - 1] + str(contador))
            contador = 1

    resultado.append(texto[-1] + str(contador))

    return "".join(resultado)


def descomprimir_texto(texto):
    if not texto:
        return ""

    resultado = []
    i = 0

    while i < len(texto):
        caracter = texto[i]

        if caracter.isdigit():
            return "ERROR: Formato comprimido inválido"

        i += 1
        inicio = i

        while i < len(texto) and texto[i].isdigit():
            i += 1

        if inicio == i:
            return "ERROR: Formato comprimido inválido"

        cantidad = int(texto[inicio:i])

        if cantidad <= 0:
            return "ERROR: Formato comprimido inválido"

        resultado.append(caracter * cantidad)

    return "".join(resultado)


comprimido = comprimir_texto(texto)
descomprimido = descomprimir_texto(comprimido)

print("Texto original:", texto)
print("Texto comprimido:", comprimido)
print("Texto descomprimido:", descomprimido)
```

**Requisitos:**
- La sala será una lista de listas donde `L` significa libre y `O` significa ocupado.
- Cada asiento solicitado se expresará mediante una tupla `(fila, columna)`, **numeradas desde 1**.
- Validar que todas las posiciones existan y que los asientos estén libres.
- Realizar la reserva únicamente si todos están disponibles.
- Evitar reservas parciales y mantener la sala intacta si ocurre un error.
- Retornar el estado, los asientos reservados y la sala actualizada.

**Reto adicional:** crea otra función `mostrar_sala(sala)` que muestre filas y columnas de manera legible.

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (3, 1)]
```

```python
# Escribe tu solución aquí
def reservar_asientos(sala, asientos_solicitados):
    for fila, columna in asientos_solicitados:
        if fila < 1 or fila > len(sala):
            return "ERROR: La fila no existe", [], sala

        if columna < 1 or columna > len(sala[fila - 1]):
            return "ERROR: La columna no existe", [], sala

        if sala[fila - 1][columna - 1] == "O":
            return f"ERROR: El asiento ({fila}, {columna}) está ocupado", [], sala

    reservados = []

    for fila, columna in asientos_solicitados:
        sala[fila - 1][columna - 1] = "O"
        reservados.append((fila, columna))

    return "Reserva realizada correctamente", reservados, sala


def mostrar_sala(sala):
    print("  1 2 3 4")

    for i, fila in enumerate(sala, start=1):
        print(i, " ".join(fila))


estado, reservados, sala_actualizada = reservar_asientos(sala, asientos)

print(estado)
print("Asientos reservados:", reservados)
mostrar_sala(sala_actualizada)
```

### Pregunta 6.1: Clase CuentaBancaria

- Método `retirar(monto)`: resta del saldo, pero debe impedir que el saldo quede negativo y avisar `"Fondos insuficientes"`.
- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# No hay datos de entrada: crea tus propios objetos para probar la clase.
```

```python
# Escribe tu solución aquí
class CuentaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, cantidad):
        if cantidad > 0:
            self.saldo += cantidad
            return "Depósito realizado correctamente"
        return "La cantidad debe ser positiva"

    def retirar(self, cantidad):
        if cantidad <= 0:
            return "La cantidad debe ser positiva"

        if cantidad > self.saldo:
            return "Fondos insuficientes"

        self.saldo -= cantidad
        return "Retiro realizado correctamente"

    def consultar_saldo(self):
        return self.saldo


cuenta = CuentaBancaria("Ana")

print("Titular:", cuenta.titular)
print(cuenta.depositar(1000))
print(cuenta.retirar(300))
print("Saldo:", cuenta.consultar_saldo())
```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
# Escribe tu solución aquí
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor
        self.disponible = True

    def prestar(self):
        if not self.disponible:
            return "El libro no está disponible"

        self.disponible = False
        return "Libro prestado correctamente"

    def devolver(self):
        self.disponible = True
        return "Libro devuelto correctamente"


class Usuario:
    def __init__(self, nombre):
        self.nombre = nombre
        self.libros_prestados = []

    def tomar_libro(self, libro):
        if not libro.disponible:
            return "El libro no está disponible"

        libro.prestar()
        self.libros_prestados.append(libro)
        return f"{self.nombre} tomó el libro '{libro.titulo}'"

    def devolver_libro(self, libro):
        if libro not in self.libros_prestados:
            return "El usuario no tiene este libro"

        libro.devolver()
        self.libros_prestados.remove(libro)
        return f"{self.nombre} devolvió el libro '{libro.titulo}'"


class Biblioteca:
    def __init__(self):
        self.libros = []
        self.usuarios = []

    def agregar_libro(self, libro):
        self.libros.append(libro)

    def registrar_usuario(self, usuario):
        self.usuarios.append(usuario)

    def mostrar_libros(self):
        for libro in self.libros:
            estado = "Disponible" if libro.disponible else "Prestado"
            print(f"{libro.titulo} - {libro.autor} - {estado}")


biblioteca = Biblioteca()

libro1 = Libro("Python básico", "Juan Pérez")
libro2 = Libro("Introducción a SQL", "Ana Torres")

usuario = Usuario("Carlos")

biblioteca.agregar_libro(libro1)
biblioteca.agregar_libro(libro2)
biblioteca.registrar_usuario(usuario)

biblioteca.mostrar_libros()

print(usuario.tomar_libro(libro1))

biblioteca.mostrar_libros()

print(usuario.devolver_libro(libro1))

biblioteca.mostrar_libros()
```
