# Programación Orientada a Objetos (POO)

```python
class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def ladrar(self):
        return f"{self.nombre} dice guau"

    def cumplir_anios(self):
        self.edad += 1
```

```python
colas = Perro("colas", 5)
```

```python
print(colas.ladrar())
```

```python
print(colas.cumplir_anios())
```

```python
print(colas.edad)
```

```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura
        
```

```python
rectangulo = Rectangulo(20, 9)
```

```python
print(f"El area del rectángulo es {rectangulo.area()}")
```

```python
class CuentaBancaria:
    def __init__(self, titular: str, saldo: float = 0.00) -> None:
        self.titular = titular
        self._saldo = saldo

    def depositar(self, monto: float):
        if monto > 0:
            self._saldo += monto

    def ver_saldo(self) -> float:
        return self._saldo
```

```python
NuevoMonto = CuentaBancaria("Marcell", 12.5)
```

```python
print(NuevoMonto.ver_saldo())
```

```python
class Animal:
    def __init__(self,  nombre: str):
        self.nombre = nombre

    def presentarse(self) -> str:
        return f"Soy {self.nombre}"
    
### MALA PRÁCTICA
class Perro(Animal):
    def ladra(self) -> str:
        return "Guau"

### BUENA PRÁCTICA

class Perro2(Animal):
    def __init__(self, nombre):
        super().__init__(nombre)
        
```

```python
boby = Perro("boby")
print(boby.ladra())
print(boby.presentarse())
```
